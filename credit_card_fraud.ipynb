# 💳 Credit Card Fraud Detection Analysis


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("credit_card_fraud_det.csv")
df.head()
```

```python
df.info()
df.describe()
```

```python
df.isnull().sum()
df.duplicated().sum()
```

## 📊 Dashboard View

### 🔹 Fraud vs Normal Transactions

```python
sns.countplot(x='is_fraud', data=df)
plt.title("Fraud vs Normal Transactions")
plt.show()
```

### 🔹 Transaction Amount Distribution

```python
sns.histplot(df['amount'], bins=40)
plt.title("Transaction Amount Distribution")
plt.show()
```

### 🔹 Fraud vs Transaction Amount

```python
sns.boxplot(x='is_fraud', y='amount', data=df)
plt.title("Fraud vs Amount")
plt.show()
```

### 🔹 Transaction Hour vs Fraud

```python
sns.countplot(x='transaction_hour', hue='is_fraud', data=df)
plt.title("Transaction Hour vs Fraud")
plt.show()
```

### 🔹 Foreign Transactions vs Fraud

```python
sns.countplot(x='foreign_transaction', hue='is_fraud', data=df)
plt.title("Foreign vs Fraud")
plt.show()
```

## 🔍 Key Insights

- Fraud transactions are rare
- Foreign transactions have higher fraud rate
- High amount transactions are risky
